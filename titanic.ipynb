```python
import pandas as pd
d_train = pd.read_csv("titanic_data/train.csv")
d_train.head()
```

```python
d_train.info()
```

```python
def extract_title(x):
    x = x.split()
    for y in x:
        if y[-1]=='.':
            return y
```

```python
d_train["Name"].map(lambda x: extract_title(x)).value_counts()
```

```python
def feature_engineering(data):
    data = data.copy()
    data.loc[:, "Sex"] = data["Sex"].map({"male":1, "female":2}).astype(int)
    data.loc[data["Age"].isna(), "Age"] = data["Age"].mode()[0]
    data.loc[data["Embarked"].isna(), "Embarked"] = data["Embarked"].mode()[0]
    data.loc[:, "Embarked"] = data["Embarked"].map({"S":1, "C":2, "Q":3})
    data.loc[:, "Family_size"] = data["SibSp"]+data["Parch"]+1
    data.loc[:, "IsAlone"] = (data["Family_size"]==1).astype(int)
    data.loc[:, "HasDeck"] = data["Cabin"].notna().astype(int)
    data.loc[:, "Title"] = data["Name"].map(lambda x: extract_title(x)).map({"Mr.": 1, "Miss.": 2, "Mrs.": 3, "Master.":4, "Dr.":5, "Rev.":6, "Mlle.":7, "Major.":7, "Col.":7, "Countess.":7, "Capt.":7, "Ms.":7, "Sir.":7, "Lady.":7, "Mme.":7, "Don.":7, "Jonkheer":7})
    data.loc[data["Title"].isna(), "Title"] = data["Title"].mode()[0].astype(int)
    data.loc[:, "Title"] = data["Title"].astype(int)
    return data
```

```python
d_train = feature_engineering(d_train)
d_train.head()
```

```python
d_train.info()
```

```python
for col in ["Sex", "Embarked"]:
    d_train[col] = pd.to_numeric(d_train[col]).astype(int)
```

```python
d_train.info()
```

```python
X = d_train.iloc[:, [2,4,5,6,7,9,11,12,13,14,15]]
Y = d_train.iloc[:, [1]]
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
```

```python
x_train,x_test, y_train, y_test = train_test_split(X,Y, train_size=0.8)
```

```python
from sklearn.metrics import accuracy_score
models = [LogisticRegression, RandomForestClassifier, GradientBoostingClassifier, XGBClassifier]
accuracy_scores = []
for model in models:
    model1 = model()
    model1.fit(x_train, y_train)
    predicts = model1.predict(x_test)
    accuracy_scores.append(accuracy_score(y_test, predicts))
print(accuracy_scores)
```

```python
comp_data = pd.read_csv("titanic_data/test.csv")
comp_data = feature_engineering(comp_data)
x_comp = comp_data.iloc[:, [1,3,4,5,6,8,10,11,12,13,14]]
x_comp.head()
```

```python
X.head()
```

```python
x_comp.loc[x_comp["Fare"].isna(), "Fare"] = x_comp["Fare"].mean()
x_comp.info()
```

```python
model = models[accuracy_scores.index(max(accuracy_scores))]()
model.fit(X,Y)
predictions = model.predict(x_comp)
predictions = pd.Series(predictions, name = "Survived", index=comp_data["PassengerId"])
predictions
```

```python
predictions.to_csv("titanic_data/output.csv")
```
